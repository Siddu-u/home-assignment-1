Q2. Manual BPE on a toy corpus

2.1 Using the same corpus from class:
low low low low low lowest lowest newer newer newer newer newer newer wider wider wider new new
Add the end-of-word marker _ and write the initial vocabulary (characters + _).
Compute bigram counts and perform the first three merges by hand:
Step 1: most frequent pair → merge → updated corpus snippet (show at least 2 lines).
Step 2: repeat.
Step 3: repeat.
After each merge, list the new token and the updated vocabulary


```python
from collections import Counter

EOW = "_"  # end-of-word marker


# ----------------------------
# 1) Build the toy corpus
# ----------------------------
# Original sequence:
# low low low low low lowest lowest newer newer newer newer newer newer wider wider wider new new
words = (
    ["low"] * 5
    + ["lowest"] * 2
    + ["newer"] * 6
    + ["wider"] * 3
    + ["new"] * 2
)

# Count each word
word_counts = Counter(words)

# Convert each unique word to a symbol list with end-of-word marker
# Example: "low" -> ["l","o","w","_"]
reps = {w: list(w) + [EOW] for w in word_counts}


def get_vocab(representations):
    """Return the set of all tokens currently used in representations."""
    vocab = set()
    for syms in representations.values():
        vocab.update(syms)
    return vocab


def get_bigram_counts(word_counts, representations):
    """Count bigrams across the corpus, weighted by word frequencies."""
    counts = Counter()
    for w, freq in word_counts.items():
        syms = representations[w]
        for a, b in zip(syms, syms[1:]):
            counts[(a, b)] += freq
    return counts


def merge_pair_in_list(symbols, pair):
    """Merge occurrences of pair (a,b) -> 'ab' in one symbol list."""
    a, b = pair
    merged = a + b
    out = []
    i = 0
    while i < len(symbols):
        if i < len(symbols) - 1 and symbols[i] == a and symbols[i + 1] == b:
            out.append(merged)
            i += 2
        else:
            out.append(symbols[i])
            i += 1
    return out, merged


def show_corpus_snippet(word_counts, representations, title, max_lines=3):
    """Print at least 2 lines showing current tokenization for common words."""
    print(title)
    shown = 0
    for w, freq in word_counts.most_common():
        print(f"  {w}_  (x{freq}):", " ".join(representations[w]))
        shown += 1
        if shown >= max_lines:
            break
    print()


# ----------------------------
# 2) Initial state
# ----------------------------
vocab0 = get_vocab(reps)
print("INITIAL VOCABULARY V0 (characters + _):")
print(sorted(vocab0))
print("Vocabulary size:", len(vocab0))
print()

show_corpus_snippet(word_counts, reps, "Initial corpus snippet (word -> symbols):", max_lines=3)


# ----------------------------
# 3) Perform first 3 merges
# ----------------------------
for step in range(1, 4):
    bigram_counts = get_bigram_counts(word_counts, reps)

    # Most frequent bigram (ties: Counter.most_common returns one of the top ties)
    best_pair, best_count = bigram_counts.most_common(1)[0]

    print(f"STEP {step}:")
    print(f"  Most frequent pair: {best_pair}  count = {best_count}")
    print(f"  Merge: {best_pair[0]} {best_pair[1]}  ->  {best_pair[0] + best_pair[1]}")

    # Apply merge to every word representation
    new_reps = {}
    new_token = None
    for w in reps:
        merged_syms, produced = merge_pair_in_list(reps[w], best_pair)
        new_reps[w] = merged_syms
        new_token = produced
    reps = new_reps

    # Show updated snippet (at least 2 lines)
    show_corpus_snippet(word_counts, reps, "  Updated corpus snippet:", max_lines=3)

    # Vocabulary update
    vocab = get_vocab(reps)
    print(f"  New token added: {new_token}")
    print(f"  Updated vocabulary size: {len(vocab)}")
    print(f"  Vocabulary (sorted): {sorted(vocab)}")
    print("\n" + "-" * 70 + "\n")
```

2.2 — Code a mini-BPE learner
Use the classroom code above (or your own) to learn BPE merges for the toy corpus.
Print the top pair at each step and the evolving vocabulary size.

```python
from collections import Counter

EOW = "_"  # end-of-word marker


# ----------------------------
# Toy corpus (same as class)
# ----------------------------
# low low low low low lowest lowest newer newer newer newer newer newer wider wider wider new new
words = (
    ["low"] * 5
    + ["lowest"] * 2
    + ["newer"] * 6
    + ["wider"] * 3
    + ["new"] * 2
)

word_counts = Counter(words)


def word_to_symbols(word: str):
    """Convert a word into a list of characters plus end-of-word marker."""
    return list(word) + [EOW]


def get_vocab(reps: dict):
    """Return the set of all tokens currently used in the representations."""
    vocab = set()
    for syms in reps.values():
        vocab.update(syms)
    return vocab


def get_bigram_counts(word_counts: Counter, reps: dict):
    """Count adjacent token pairs weighted by word frequency."""
    pair_counts = Counter()
    for w, freq in word_counts.items():
        syms = reps[w]
        for a, b in zip(syms, syms[1:]):
            pair_counts[(a, b)] += freq
    return pair_counts


def merge_pair(symbols: list, pair: tuple):
    """Merge occurrences of (a,b) into a single token 'ab'."""
    a, b = pair
    merged_token = a + b

    out = []
    i = 0
    while i < len(symbols):
        # If current and next symbol form the pair, merge them
        if i < len(symbols) - 1 and symbols[i] == a and symbols[i + 1] == b:
            out.append(merged_token)
            i += 2
        else:
            out.append(symbols[i])
            i += 1

    return out, merged_token


def learn_bpe(word_counts: Counter, num_merges: int = 10):
    """
    Learn BPE merges.
    Prints:
      - top pair each step
      - evolving vocabulary size
    Returns:
      - merges list (in order)
      - final reps
      - final vocab set
    """
    # Initial word representations (characters + _)
    reps = {w: word_to_symbols(w) for w in word_counts.keys()}

    # Initial vocab
    vocab = get_vocab(reps)
    print("Initial vocabulary size:", len(vocab))
    print("Initial vocabulary:", sorted(vocab))
    print()

    merges = []

    for step in range(1, num_merges + 1):
        pair_counts = get_bigram_counts(word_counts, reps)
        if not pair_counts:
            break

        best_pair, best_count = pair_counts.most_common(1)[0]

        # Apply merge to each word representation
        new_reps = {}
        new_token = None
        for w in reps:
            merged_syms, produced = merge_pair(reps[w], best_pair)
            new_reps[w] = merged_syms
            new_token = produced

        reps = new_reps
        merges.append(best_pair)

        # Update vocab
        vocab = get_vocab(reps)

        print(f"Step {step:02d}: best pair = {best_pair}, count = {best_count}")
        print(f"          new token = {new_token}, vocab size = {len(vocab)}")
        print()

    return merges, reps, vocab


def apply_bpe_to_word(word: str, merges: list):
    """
    Segment a new word by applying learned merges in order.
    Returns list of subword tokens (including _ where applicable).
    """
    symbols = word_to_symbols(word)

    for pair in merges:
        a, b = pair
        merged_token = a + b

        out = []
        i = 0
        while i < len(symbols):
            if i < len(symbols) - 1 and symbols[i] == a and symbols[i + 1] == b:
                out.append(merged_token)
                i += 2
            else:
                out.append(symbols[i])
                i += 1
        symbols = out

    return symbols


if __name__ == "__main__":
    # Learn merges (you can change 10 to any number you want)
    merges, reps, vocab = learn_bpe(word_counts, num_merges=10)

    print("Learned merges (in order):")
    for i, (a, b) in enumerate(merges, 1):
        print(f"{i:02d}. ({a}, {b}) -> {a + b}")
    print()

    # Required segmentations
    test_words = ["new", "newer", "lowest", "widest", "newestest"]  # invented word
    print("Segmentation results:")
    for w in test_words:
        seg = apply_bpe_to_word(w, merges)
        print(f"{w} -> {' '.join(seg)}")
```

```python
import re
from collections import Counter

EOW = "_"  # end-of-word marker


def preprocess(text: str):
    """
    Lowercase + extract word tokens (letters + apostrophes).
    """
    return re.findall(r"[A-Za-z']+", text.lower())


def init_vocab(words):
    """
    Build:
      - counts: word -> frequency
      - reps: word -> list of symbols (characters + _)
    """
    counts = Counter(words)
    reps = {w: list(w) + [EOW] for w in counts}
    return counts, reps


def get_vocab(reps: dict):
    """Current vocabulary = all tokens used in reps."""
    vocab = set()
    for syms in reps.values():
        vocab.update(syms)
    return vocab


def get_pair_counts(counts: Counter, reps: dict):
    """Count adjacent pairs weighted by word frequency."""
    pc = Counter()
    for w, freq in counts.items():
        syms = reps[w]
        for a, b in zip(syms, syms[1:]):
            pc[(a, b)] += freq
    return pc


def merge_pair(symbols: list, pair: tuple):
    """Merge occurrences of (a,b) into a single token 'ab'."""
    a, b = pair
    merged = a + b

    out = []
    i = 0
    while i < len(symbols):
        if i < len(symbols) - 1 and symbols[i] == a and symbols[i + 1] == b:
            out.append(merged)
            i += 2
        else:
            out.append(symbols[i])
            i += 1

    return out, merged


def learn_bpe(text: str, num_merges: int = 30):
    """
    Learn BPE merges from text.
    Prints the top pair each step + evolving vocabulary size.
    Returns merges, merge_log, final vocab.
    """
    words = preprocess(text)
    counts, reps = init_vocab(words)

    merges = []
    merge_log = []  # (pair, count_at_merge, new_token)

    vocab = get_vocab(reps)

    for step in range(1, num_merges + 1):
        pair_counts = get_pair_counts(counts, reps)
        if not pair_counts:
            break

        best_pair, best_count = pair_counts.most_common(1)[0]

        # apply merge to all word reps
        new_reps = {}
        new_token = None
        for w in reps:
            merged_syms, produced = merge_pair(reps[w], best_pair)
            new_reps[w] = merged_syms
            new_token = produced

        reps = new_reps
        merges.append(best_pair)
        merge_log.append((best_pair, best_count, new_token))

        vocab = get_vocab(reps)

        print(f"Step {step:02d}: best pair = {best_pair}, count = {best_count}, vocab size = {len(vocab)}")

    return merges, merge_log, vocab


def apply_bpe_to_word(word: str, merges: list):
    """
    Segment a word by applying merges in order.
    Returns token list including _ where applicable.
    """
    symbols = list(word.lower()) + [EOW]

    for a, b in merges:
        out = []
        i = 0
        merged = a + b
        while i < len(symbols):
            if i < len(symbols) - 1 and symbols[i] == a and symbols[i + 1] == b:
                out.append(merged)
                i += 2
            else:
                out.append(symbols[i])
                i += 1
        symbols = out

    return symbols


if __name__ == "__main__":
    paragraph = (
        "Natural language processing is messy, but tokenization makes it manageable. "
        "In this class we compare words, subwords, and characters, because real text contains typos and new terms. "
        "A good tokenizer learns frequent patterns such as learning, learned, and learner. "
        "Rare words like electroencephalography still become representable with subwords. "
        "That is why subword methods improve generalization and efficiency."
    )

    print("\n--- Learn at least 30 merges ---")
    merges, merge_log, vocab = learn_bpe(paragraph, num_merges=30)

    # Five most frequent merges (highest count at the moment of merging)
    top5_merges = sorted(merge_log, key=lambda x: x[1], reverse=True)[:5]
    print("\n--- Five most frequent merges (pair -> new_token, count) ---")
    for (pair, ct, tok) in top5_merges:
        print(f"{pair} -> {tok}   (count={ct})")

    # Five longest tokens in the final vocabulary
    longest5 = sorted(vocab, key=len, reverse=True)[:5]
    print("\n--- Five longest subword tokens (token, length) ---")
    for tok in longest5:
        print(f"{tok}  (len={len(tok)})")

    # Segment 5 words from the paragraph:
    # - include one rare word: electroencephalography
    # - include one derived/inflected form: generalization, learned
    test_words = ["electroencephalography", "generalization", "learning", "learned", "tokenization"]

    print("\n--- Segmentation of 5 words ---")
    for w in test_words:
        print(w, "->", " ".join(apply_bpe_to_word(w, merges)))
```
